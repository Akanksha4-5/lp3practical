```python
df = df.drop(['RowNumber', 'Surname', 'CustomerId'], axis= 1) #Dropping the columns that don't help prediction
```

```python
df.head()
```

```python
def visualization(x, y, xlabel):
    plt.figure(figsize=(10,5))
    plt.hist([x, y], color=['red', 'green'], label = ['exit', 'not_exit'])
    plt.xlabel(xlabel,fontsize=20)
    plt.ylabel("No. of customers", fontsize=20)
    plt.legend()
```

```python
df_churn_exited = df[df['Exited']==1]['Tenure']
df_churn_not_exited = df[df['Exited']==0]['Tenure']
```

```python
visualization(df_churn_exited, df_churn_not_exited, "Tenure")
```

```python
df_churn_exited2 = df[df['Exited']==1]['Age']
df_churn_not_exited2 = df[df['Exited']==0]['Age']
```

```python
visualization(df_churn_exited2, df_churn_not_exited2, "Age")
```

```python
states = pd.get_dummies(df['Geography'],drop_first = True, dtype=int)
gender = pd.get_dummies(df['Gender'],drop_first = True, dtype=int)
```

```python
df = pd.concat([df,gender,states], axis = 1)
```

```python
df.head()
```

```python
X = df[['CreditScore','Age','Tenure','Balance','NumOfProducts','HasCrCard','IsActiveMember','EstimatedSalary','Male','Germany','Spain']]
```

```python
y = df['Exited']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.30, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
X_train  = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
import keras #Keras is the wrapper on the top of tensorflow
```

```python
from keras.models import Sequential #To create sequential neural network
from keras.layers import Dense #To create hidden layers
```

```python
classifier = Sequential()
```

```python
#To add the layers
#Dense helps to construct the neurons
#Input Dimension means we have 11 features
# Units is to create the hidden layers
#Uniform helps to distribute the weight uniformly
classifier.add(Dense(activation = "relu",input_dim = 11,units = 6,kernel_initializer = "uniform"))
```

```python
classifier.add(Dense(activation = "relu",units = 6,kernel_initializer = "uniform"))
```

```python
classifier.add(Dense(activation = "sigmoid",units = 1,kernel_initializer = "uniform"))
```

```python
classifier.compile(optimizer="adam",loss = 'binary_crossentropy',metrics = ['accuracy'])
```

```python
classifier.summary() # 6 neurons in 1st, 6 neurons in 2nd layer and 1 neuron in last
```

```python
classifier.fit(X_train,y_train,batch_size=10,epochs=50) #Fitting the ANN to training dataset
```

```python
y_pred =classifier.predict(X_test)
y_pred = (y_pred > 0.5) #Predicting the result
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
cm
```

```python
accuracy = accuracy_score(y_test,y_pred)
```

```python
accuracy
```

```python
plt.figure(figsize = (10,7))
sns.heatmap(cm,annot = True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
print(classification_report(y_test,y_pred))
```
